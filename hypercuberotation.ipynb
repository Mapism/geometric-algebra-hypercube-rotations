```python
import kingdon
from kingdon import Algebra
import numpy as np
import sympy as sp
import math
from random import random as rand
from itertools import product
from kingdon import *


def exp(x):
    l = (x*x).e
    
    if l > 0:
        lsqrt = np.sqrt(l)
        a = np.cosh(lsqrt) + (np.sinh(lsqrt) / lsqrt) * x
        return a
    if l < 0:
        lsqrt = np.sqrt(-l)
        a = np.cos(lsqrt) + (np.sin(lsqrt) / lsqrt) * x 
        return a
    return 1 + x
```

```python
N = int(input("Enter dimension:")) 
algGA = Algebra(N, 0, 0)
locals().update(algGA.blades)
angle = float(input("Enter angle:"))


def vertices(N): # coordinates of vertices of N dimensional cube
    return list(product((0, 1), repeat=N))
coordi = vertices(N)
#print(coord)
#print(" ")


def ga_vert(N): # gavectors representing vertices of N dimensional cube
    l1 = [0]
    for i in range (1, 2**(N)):
        s=0
        for j in range (0, N):
            s=s+coordi[i][j]*(algGA.blades["e"+ str(j+1)])
        l1.append(s)
    return (l1)
ivertGA = ga_vert(N)
#for i in range(0, 2**N): print(lGA[i]), print(" ")
#print(ivertGA[3].values())


def initialproject(algGA): # initial gavectors(cube) projection onto 3D
    I = e1*e2*e3
    l1 = [0]
    for i in range(1, 2**N):
        s = (ivertGA[i].lc(I))*e3*e2*e1
        s = s.grade(1)
        l1.append(s)
    return l1
iProjGA = initialproject(algGA)
#for i in range(0, 2**N): print(iProjGA[i]), print(" ")


def initialcoeff(algGA): # converting initial projected gavectors representing the vertices of N dimensional cube to coordinates
    l1 = [[0,0,0]]
    for i in range(1, 2**N):
        x = iProjGA[i]
        x = x.map(lambda v: round(v, 3))
        t1 = list(x.values())
        l1.append(t1) #adding t1 to l1(our list of final coordinates)
    return l1
iProjcoord = initialcoeff(algGA)
#print(iProjcoord)
```

```python
def garotor(algGA):
    plane = e1*e2+e2*e4+e4*e3+2*e2*e4            #3*e1*e6+e2*e3+e1*e3+2*e4*e6 #I manually set this as the plane of rotation  e1*e2+e2*e4+e1*e3+e2*e3+e4*e3
    plane = plane.normalized()
    R = exp((plane)*angle*math.pi/180/2)
    R = R.map(lambda v: round(v, 3))
    return R
R = garotor(algGA)
#print(R)
#print(R.values())
#print(garotor(algGA)) R = garotor print(R.values())


# def randomgarotor(algGA):
#     x=0
#     y=0
#     for i in range(0,N):
#         x = x+rand()*(algGA.blades["e"+ str(i+1)])
#         y = y+rand()*(algGA.blades["e"+ str(i+1)])
#     R = x*y/x.norm()/y.norm()
#     R = R.map(lambda v: round(v, 3))
#     return R
# R = randomgarotor(algGA)
# #print(randomGA_rotor(algGA)), print(" ")


def rotate(algGA): # rotation of gavectors representing vertices of N dimensional cube
    l1 = [0]
    for i in range(1, 2**N):
        s = R.reverse()*ivertGA[i]
        s = s*R
        s = s.map(lambda v: round(v, 3))
        #s = s.normalized() 
        l1.append(s)
    return l1
Rot = rotate(algGA)
#print("Rot")
#for i in range(0, 2**N): print(Rot[i])
#print(" ")


def finalproject(algGA): # rotated gavectors representing vertices of N dimensional cube projection onto 3D
    I = e1*e2*e3
    l1 = [0]
    for i in range(1, 2**N):
        s = (Rot[i].lc(I))*e3*e2*e1
        s = s.grade(1)
        s = s.map(lambda v: round(v, 3))
        l1.append(s)
    return l1
fProjGA = finalproject(algGA)
#print(fProjGA[3])
#print(fProjGA[3].values())
#for i in range(0, 2**N): print(finalProj[i]), print(" ")


def listofvalues(): # tuple of values
    l1=[(0,0,0)]
    for i in range (1,len(fProjGA)):
        a = fProjGA[i].values()
        l1.append(a)
    return l1
listvalues=listofvalues()
#print(listvalues)


def roundlist(): # rounded down list of values
    l1 = []
    for i in range (0,2**N):
        l2 = []
        for j in range (0, 3):
            l2.append(round(listvalues[i][j],3))
        l1.append(l2)
    return l1
fProjcoord = roundlist()
#print(fProjcoord)


# def finalcoeff(algGA): # converting final projected gavectors representing the vertices of N dimensional cube to coordinates
#     l1 = [[0,0,0]]
#     for i in range(1, 2**N):
#         t1 = list(roundl[i])
#         l1.append(t1) #adding t1 to l1(our list of final coordinates)
#     return l1
# fProjcoord = finalcoeff(algGA)
# print(fProjcoord)
```

```python
algPGA = Algebra(3, 0, 1)
locals().update(algPGA.blades)

point = lambda x, y, z: (algPGA.blades["e0"] + x*algPGA.blades["e1"] + y*algPGA.blades["e2"] + z*algPGA.blades["e3"]).dual()


def graphi(algPGA): # changing the initial vertice coordinates to PGA points
    l1 = [point(0,0,0)]
    for i in range(1, 2**N):
        A = point(coordi[i][0],coordi[i][1],coordi[i][2])
        l1.append(A)
    return l1
pointsi = graphi(algPGA)
#print(pointsi[3].values())


def indicei(N): #!
    l1 = []
    for i in range(2**N):
        for j in range(0, N):
            ind2 = i^(2**j)
            if(ind2<i):
                continue
            b=[pointsi[i],pointsi[ind2]]
            l1.append(b)
    return l1
linesi = indicei(N)
#print(linesi[3])





def graphf(algPGA): # changing the final vertice coordinates to PGA points
    l1 = [point(0,0,0)]
    for i in range(1, 2**N):
        A = point(fProjcoord[i][0],fProjcoord[i][1],fProjcoord[i][2])
        l1.append(A)
    return l1
pointsf = graphf(algPGA)
#print(pointsf[3].values())


def indicef(N): #!
    l1 = []
    for i in range(2**N):
        for j in range(0, N):
            ind2 = i^(2**j)
            if(ind2<i):
                continue
            b=[pointsf[i],pointsf[ind2]]
            l1.append(b)
    return l1
linesf = indicef(N)
#print(linesf[3])


display(algPGA.graph(
    0x870f0f, e1*e2,"z", e2*e3,"x", e3*e1,"y",                    # vector bases
    #0xD0FFE1, [A,B,D],                                           # polygons
    0x0a8068,  *linesi,                                          # edgesi
    #0x0a8068,  pointsi,                                           # pointsi
    0x0703fc, *linesf,                                           # edgesf
    #0x224488, pointsf,                                            # pointsf
    
    
    lineWidth=3,
    grid=1,
    labels=1,
    h=0.6,
    p=-0.15,
    pointRadius=1,
    fontSize=1,
    scale=1,
))
```
